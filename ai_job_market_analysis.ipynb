## 6. Dataset Source and Collection

The dataset was published on Kaggle by Maria Aqdas. The job posting data was collected through the Adzuna Jobs API and includes job postings from the United States, United Kingdom, and India.

The dataset contains job postings from 2026. The data was gathered using job-search keywords, including terms related to artificial intelligence and automation. Because the information comes from online job postings available through the Adzuna API, the dataset represents jobs that were available through that source during the collection period rather than every job available in these countries.

## 7. Possible Biases

There are some possible biases in this dataset. One is that the job postings only come from the United States, United Kingdom, and India, so the dataset does not represent the entire global job market. Certain countries and types of jobs may therefore be overrepresented or underrepresented.

Another possible bias comes from the way the jobs were collected. Since the data came through the Adzuna Jobs API and specific search keywords were used, jobs that were not available through Adzuna or did not match those keywords may have been left out. I also noticed that some job postings are missing information such as salary or contract type. Because of these limitations, I would be careful about using this dataset to make conclusions about the entire job market.
